<h3 style='color:green'>Project Overview</h3

**This project analyzes a student dataset containing academic and lifestyle-related factors such as age, study hours, sleep duration, CGPA, stress levels, attendance percentage, and social media usage. The main objective is to understand how these variables influence student performance and to apply statistical and linear algebra techniques for extracting insights.**

**Aims & Objectives**

• To preprocess the dataset by handling missing values, imputations, and duplicates.

• To perform exploratory data analysis (EDA) and understand the distribution and relationships between variables.

• To apply Principal Component Analysis (PCA) for dimensionality reduction and identify which components capture the maximum variance.

• To apply Linear Regression to examine the relationship between study hours and CGPA.

• To derive meaningful inferences that can help explain how different factors (study habits, stress, attendance, etc.) correlate with student performance.

**Scope**

• Through this project, we aim to:

• Prepare a clean and analysis-ready dataset.

• Use PCA to reduce complexity while retaining most of the information.

• Use regression to explore predictive relationships between features.

• Provide insights into the factors influencing students’ academic performance.

<h3 style='color:green'>Exploratory Data Analysis (EDA)</h3>

**CGPA Distribution (Histogram + KDE)**

• Observation: Most CGPAs lie roughly between 6 and 8.

• Inference: Majority perform in the average-to-good band; a small tail of high performers.

**Attendance vs CGPA (Scatter, colored by Gender)**

• Observation: A mild positive trend higher attendance often aligns with higher CGPA; points for different genders largely overlap.

• Inference: Attendance likely helps performance; no visible gender gap in this sample.

**CGPA vs Stress Level (Boxplot)**

• Observation: Lower stress (levels 1–2) shows slightly higher median CGPA; distributions across stress groups overlap considerably.

•Inference: Stress may have a negative association with performance, but effect appears small/uncertain.

**Bar Plot of Average CGPA by Stress Level**

• Observation: The average CGPA is highest at stress level 1 and lowest at stress level 2. Stress levels 3, 4, and 5 show similar average CGPAs, with no major differences.

• Inference: Lower stress levels are linked to better academic performance.

**Line Plot: Study Hours vs Average CGPA**

• Observation: The trend line between study hours and CGPA shows high fluctuations, with students achieving a wide range of CGPAs regardless of how many hours they study per week.

• Inference: This suggests that study hours alone are not a strong predictor of academic performance; other factors like study methods, stress, and consistency may play a bigger role.

<h4 style='color:green'>Statistical Testing</h4>

**Part-Time Job vs CGPA (Independent t-test)**

Result: t ≈ −0.11, p ≈ 0.91 (using nan_policy='omit').

Inference: No statistically significant difference in CGPA between students with and without part-time jobs in this sample.

We compared CGPA between students with and without part-time jobs. The test showed no significant difference (p ≈ 0.91), meaning part-time jobs did not affect academic performance in this sample.

**Stress Level Groups vs CGPA (One-Way ANOVA)**

• Result: F ≈ 1.84, p ≈ 0.17 (across stress groups tested).

• Inference: Fail to reject equal means, no significant CGPA differences across stress levels, though EDA suggested a small trend.

We compared CGPA across students with low, medium, and high stress levels. The ANOVA results (p ≈ 0.16) indicated no statistically significant difference, suggesting that stress level alone does not strongly impact CGPA.

<h4 style='color:purple'>Principal Component Analysis</h4>

**Objective**:

• To reduce the dimensionality of the dataset and identify the principal components (PCs) that capture most of the variance in the features.

Methodology:

• Selected numeric features:

• Age

• Study hours per week

• Sleep hours

• CGPA

• Stress level

• Attendance percentage

• Social media hours

• Standardized the dataset using StandardScaler (important so all features contribute equally).

• Applied PCA using sklearn.decomposition.PCA.

• Extracted explained variance ratios and cumulative variance.

Results:

• Explained Variance Ratio:
[0.193, 0.183, 0.162, 0.143, 0.129, 0.101, 0.089]

• Cumulative Variance Explained:
[0.193, 0.377, 0.538, 0.681, 0.810, 0.911, 1.000]

**Observations**:

• The first principal component (PC1) explains ~19.3% of the variance.

• The first 3 PCs together explain ~53.8% of the total variance.

• The first 5 PCs together explain ~81% of the variance.

• All 7 PCs capture 100% (since no information is lost if all are kept).

**Inference**:

PCA shows that the dataset’s variance is spread across multiple components rather than being dominated by a single one.

Dimensionality reduction is possible: instead of keeping all 7 features, we can keep the top 5 PCs and still retain ~81% of the dataset’s information.

This highlights that features like study hours, stress, attendance, etc., interact in complex ways to influence student performance.


<h4 style='color:purple'>Linear Regression Analysis</h4>

**Objective**: To examine whether study hours per week can be used as a predictor for CGPA.

**Methodology**:

•  Selected CGPA as the dependent variable (y).
    
• Applied Linear Regression using sklearn.linear_model.LinearRegression.

• Fitted the model, obtained regression coefficients, and visualized the regression line.

**Results**:

• Slope: ~-0.0000065

• Regression line is almost flat, suggesting negligible effect of study hours on CGPA.

• Scatter plot shows a high spread of actual data points around the regression line.

**Observation**:

• The regression line does not fit the data well.

• Increasing study hours per week does not show a strong or consistent impact on CGPA.

**Inference**:

• There is no significant linear relationship between study hours and CGPA.

• Students’ CGPA seems to depend on multiple other factors beyond just study hours (like study quality, stress, sleep, and lifestyle).

<h3 style='color:green'>Conclusion</h3>

The analysis of the student dataset provided valuable insights into the factors influencing academic performance. Through data preprocessing (handling missing values, imputations, and duplicates), the dataset was prepared for reliable analysis.

Overall, the study suggests that student performance (CGPA) cannot be explained by a single factor like study hours, stress, or part-time work. Instead, it is influenced by a complex combination of factors, as revealed by PCA.
